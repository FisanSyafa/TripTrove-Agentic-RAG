Notebook untuk testing dan debugging RAG system secara interaktif.

## Tujuan:
- Test retrieval quality
- Debug agent responses
- Analyze vector search results
- Measure response time

```python
# Setup path
import sys
from pathlib import Path

project_root = Path.cwd().parent
src_path = project_root / 'src'
sys.path.insert(0, str(src_path))

print(f"✅ Path configured: {src_path}")
```

```python
# Import libraries
from agent_rag import TripTroveAgent
from data_loader import TripTroveDataLoader
import time
from datetime import datetime

print("✅ Libraries imported!")
```

## 1. Initialize Agent

```python
# Initialize agent
print("🔄 Initializing TripTrove Agent...")
agent = TripTroveAgent()
print("✅ Agent initialized!")
```

## 2. Test Vector Search

```python
# Test retrieval
test_query = "paket tour ke Bali"

print(f"🔍 Testing retrieval for: '{test_query}'\n")
docs = agent.vectorstore.similarity_search(test_query, k=5)

print(f"📄 Retrieved {len(docs)} documents:\n")
for i, doc in enumerate(docs, 1):
    print(f"Document {i}:")
    print(f"Content: {doc.page_content[:200]}...")
    print(f"Metadata: {doc.metadata}")
    print("-" * 80)
```

## 3. Test Agent Queries

```python
# Test queries
test_queries = [
    "Paket tour apa saja yang tersedia?",
    "Berapa harga paket tour ke Bali?",
    "Apa term and condition TripTrove?",
    "Rekomendasi paket tour untuk keluarga?",
    "Paket tour dengan budget 5 juta?"
]

results = []

for query in test_queries:
    print(f"\n{'='*80}")
    print(f"❓ Query: {query}")
    print(f"{'='*80}")
    
    start_time = time.time()
    response = agent.query(query)
    elapsed = time.time() - start_time
    
    print(f"\n🤖 Response:\n{response}")
    print(f"\n⏱️ Time: {elapsed:.2f}s")
    
    results.append({
        'query': query,
        'response': response,
        'time': elapsed
    })
```

```python
# Summary statistics
import pandas as pd

df_results = pd.DataFrame(results)
print("\n📊 Performance Summary:")
print(f"Average response time: {df_results['time'].mean():.2f}s")
print(f"Min response time: {df_results['time'].min():.2f}s")
print(f"Max response time: {df_results['time'].max():.2f}s")
```

## 4. Interactive Testing

```python
# Interactive query (run this cell multiple times with different queries)
query = input("Enter your query: ")

if query:
    print(f"\n🔍 Processing: {query}\n")
    
    start = time.time()
    response = agent.query(query)
    elapsed = time.time() - start
    
    print(f"🤖 Response:\n{response}")
    print(f"\n⏱️ Time: {elapsed:.2f}s")
```

## 5. Debug Mode - Inspect Retrieval

```python
# Debug specific query
debug_query = "term and condition"

print(f"🔍 Debug Query: '{debug_query}'\n")

# Get retrieved documents
docs = agent.vectorstore.similarity_search(debug_query, k=8)

print(f"📄 Retrieved {len(docs)} documents:\n")

for i, doc in enumerate(docs, 1):
    print(f"\n{'='*80}")
    print(f"Document {i}")
    print(f"{'='*80}")
    print(f"Source: {doc.metadata.get('source', 'unknown')}")
    print(f"Type: {doc.metadata.get('source_type', 'unknown')}")
    print(f"\nContent:\n{doc.page_content}")
```
