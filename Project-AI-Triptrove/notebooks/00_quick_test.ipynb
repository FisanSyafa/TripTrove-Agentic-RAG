```python
from langchain_ollama import ChatOllama

# Mengkoneksikan kode ke Ollama lokal Anda
llm = ChatOllama(model="llama3.1")

print("Meminta jawaban dari AI Lokal...")

# Memberikan perintah ke AI
response = llm.invoke("Jawab dengan singkat: Apa keuntungan menjalankan AI secara lokal?")

print("\nJawaban AI:")
print(response.content)
```
